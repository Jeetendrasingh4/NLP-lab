**Punkt** is a popular, unsupervised pre-trained tokenizer within the Python Natural Language Toolkit (NLTK) library, designed to split raw text into sentences and words.

```python
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize

# Example text
#text = """NLTK is a leading platform for building Python programs to work with human language data.
#It provides easy-to-use interfaces to over 50 corpora and lexical resources such as WordNet,
#along with a suite of text processing libraries for classification, tokenization, stemming,
#tagging, parsing, and semantic reasoning, wrappers for industrial-strength NLP libraries,
#and an active discussion forum."""

text = '''Dr. A.P.J. Abdul Kalam was the 11th President of India and a renowned aerospace scientist.
He was born on October 15, 1931, in Rameswaram, Tamil Nadu.
Kalam played a key role in India's Pokhran-II nuclear tests in 1998.
He worked with ISRO and DRDO on missile development programs such as Agni and Prithvi.
In 2002, he became the President of India and served until 2007.
He received the Bharat Ratna, India's highest civilian award, in 1997.
Dr. Kalam inspired millions of students and delivered lectures at universities across the world.'''

# Tokenize sentences in paragraph
sentences = sent_tokenize(text)

# Print tokenized sentences
for i, sentence in enumerate(sentences):
    print(f"Sentence {i+1}: {sentence}")
```

```python
from nltk.tokenize import word_tokenize

# Tokenize words in sentences
for i, sentence in enumerate(sentences):
    words = word_tokenize(sentence)
    print(f"Sentence {i+1}: {words}")
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords

# Get English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words from tokenized sentences
filtered_sentences_words = []
for sentence_words in [word_tokenize(s) for s in sentences]:
    filtered_words = [word for word in sentence_words if word.lower() not in stop_words]
    filtered_sentences_words.append(filtered_words)

# Print sentences after stop word removal
print("\nSentences after stop word removal:")
for i, filtered_words in enumerate(filtered_sentences_words):
    print(f"Sentence {i+1}: {filtered_words}")
```

```python
from nltk.stem import PorterStemmer

# Initialize Porter Stemmer
porter_stemmer = PorterStemmer()

# Perform stemming on the filtered words
stemmed_sentences_words = []
for sentence_words in filtered_sentences_words:
    stemmed_words = [porter_stemmer.stem(word) for word in sentence_words]
    stemmed_sentences_words.append(stemmed_words)

# Print sentences after stemming
print("\nSentences after stemming:")
for i, stemmed_words in enumerate(stemmed_sentences_words):
    print(f"Sentence {i+1}: {stemmed_words}")
```

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer

# Initialize WordNet Lemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

# Perform lemmatization on the filtered words
lemmatized_sentences_words = []
for sentence_words in filtered_sentences_words:
    lemmatized_words = [wordnet_lemmatizer.lemmatize(word) for word in sentence_words]
    lemmatized_sentences_words.append(lemmatized_words)

# Print sentences after lemmatization
print("\nSentences after lemmatization:")
for i, lemmatized_words in enumerate(lemmatized_sentences_words):
    print(f"Sentence {i+1}: {lemmatized_words}")
```

### Comparison of Stemming and Lemmatization

Here, we compare the words after stop word removal, stemming, and lemmatization for each sentence. Notice how stemming often chops off suffixes to get to a root form that might not be a valid word, while lemmatization aims to get to a dictionary-valid base form (lemma).

```python
print("\n--- Comparison of Text Processing ---\n")
for i in range(len(filtered_sentences_words)):
    print(f"Sentence {i+1}:")
    print(f"  Original (filtered): {filtered_sentences_words[i]}")
    print(f"  Stemmed:             {stemmed_sentences_words[i]}")
    print(f"  Lemmatized:          {lemmatized_sentences_words[i]}\n")
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
from nltk import pos_tag

# Perform POS tagging on the filtered words
tagged_sentences_words = []
for sentence_words in filtered_sentences_words:
    tagged_words = pos_tag(sentence_words)
    tagged_sentences_words.append(tagged_words)

# Print sentences after POS tagging
print("\nSentences after POS Tagging:")
for i, tagged_words in enumerate(tagged_sentences_words):
    print(f"Sentence {i+1}: {tagged_words}")
```

```python
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('maxent_ne_chunker_tab') # Download the missing resource
from nltk import ne_chunk

# Perform Named Entity Recognition (NER)
named_entities_sentences = []
for tagged_words in tagged_sentences_words:
    named_entities = ne_chunk(tagged_words)
    named_entities_sentences.append(named_entities)

# Print sentences with Named Entities
print("\nSentences with Named Entity Recognition:")
for i, named_entities in enumerate(named_entities_sentences):
    print(f"Sentence {i+1}: {named_entities}")
```

```python
# Install stanza
!pip install stanza

import stanza

# Download English model
stanza.download('en')

# Load English pipeline with tokenize processor
nlp = stanza.Pipeline(lang='en', processors='tokenize')

# Example text
text = """Natural Language Processing is a subfield of Artificial Intelligence.
It deals with interaction between computers and human language."""

# Process text
doc = nlp(text)

# Sentence Tokenization
print("----- Sentence Tokenization (Stanza) -----")
for i, sentence in enumerate(doc.sentences):
    print(f"Sentence {i+1}: {sentence.text}")

# Word Tokenization
print("\n----- Word Tokenization (Stanza) -----")
for i, sentence in enumerate(doc.sentences):
    words = [word.text for word in sentence.words]
    print(f"Sentence {i+1} Words: {words}")
```

```python
# ================= TF-IDF =================
from sklearn.feature_extraction.text import TfidfVectorizer

# Use same sentences already generated
documents = [" ".join(sentence) if isinstance(sentence, list) else sentence for sentence in sentences]

# Preprocess (optional: use filtered words)
processed_docs = [" ".join([w.lower() for w in word_tokenize(doc) if w.isalnum()]) for doc in documents]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(processed_docs)

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

#Output: a list of all unique words (vocabulary)
print("\nFeature Names:")
print(vectorizer.get_feature_names_out())
```

```python
# ================= CYK / Chart Parsing =================
import nltk
from nltk import CFG
from nltk.parse import ChartParser

# Define grammar
grammar = CFG.fromstring("""
S -> NP VP
VP -> V NP
NP -> Det N
Det -> 'the' | 'a'
N -> 'cat' | 'dog'
V -> 'chased' | 'saw'
""")

parser = ChartParser(grammar)

sentence = "the cat chased the dog".split()

print("\nCYK / Chart Parsing Trees:")
for tree in parser.parse(sentence):
    print(tree)
```

```python
# ================= N-GRAMS =================
from nltk.util import ngrams

sample_text = "this is a simple nlp example for ngram generation"
tokens = word_tokenize(sample_text)

print("\nUnigrams:")
print(list(ngrams(tokens, 1)))

print("\nBigrams:")
print(list(ngrams(tokens, 2)))

print("\nTrigrams:")
print(list(ngrams(tokens, 3)))
```

```python
# ================= SENTIMENT ANALYSIS =================
from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()

text_samples = [
    "I love this product! It is amazing.",
    "This is the worst experience I have ever had.",
    "The movie was okay, not great but not bad."
]

print("\nSentiment Analysis Results:")
for text in text_samples:
    score = sia.polarity_scores(text)
    print(f"\nText: {text}")
    print(f"Scores: {score}")

    if score['compound'] >= 0.05:
        print("Sentiment: Positive")
    elif score['compound'] <= -0.05:
        print("Sentiment: Negative")
    else:
        print("Sentiment: Neutral")
```
